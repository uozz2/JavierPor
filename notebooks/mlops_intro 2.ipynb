## 🔍 Sesión 2: Gestión de Experimentos con MLflow

````python
import mlflow
import mlflow.sklearn
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.datasets import load_wine

# Cargar datos
data = load_wine()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2)

# Iniciar experimento
mlflow.set_experiment("MLOps Intro")

with mlflow.start_run():
    # Entrenar modelo
    model = RandomForestClassifier(n_estimators=100)
    model.fit(X_train, y_train)
    
    # Evaluar
    preds = model.predict(X_test)
    acc = accuracy_score(y_test, preds)
    
    # Registrar parámetros y métricas
    mlflow.log_param("n_estimators", 100)
    mlflow.log_metric("accuracy", acc)
    
    # Guardar modelo
    mlflow.sklearn.log_model(model, "model")
````

Después de ejecutar esto, inicia la interfaz web de MLflow en consola:
```
mlflow ui
```
Y visita: http://localhost:5000

## 🚀 Sesión 3: Despliegue Básico con FastAPI

Crea un archivo `main.py` con este contenido:

````python
from fastapi import FastAPI
import joblib
import numpy as np

app = FastAPI()
model = joblib.load("wine_model.pkl")

@app.post("/predict")
def predict(data: list):
    prediction = model.predict([data])
    return {"prediction": int(prediction[0])}
````

Ejecuta desde consola:
```
uvicorn main:app --reload
```
Visita la documentación interactiva en:
[http://localhost:8000/docs](http://localhost:8000/docs)
